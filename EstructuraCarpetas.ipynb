```python
from os.path import join
from pathlib import Path
from os import makedirs
import getpass
```

```python
# ============================================================
# 1. MAIN PATH
# ============================================================

username = getpass.getuser()

if username == "JOEL":
    main_path = Path("D:/Github/PE1333_Grupo8")

elif username == "USUARIO_alterno":
    main_path = Path("C:/Users/USUARIO_alterno/....")

print(f"Directorio del proyecto: {main_path}")
```

```python
# ============================================================
# 2. DEFINIR DIRECTORIOS PRINCIPALES
# ============================================================

tp = 1

a_path = join(main_path, "TP_0"+str(tp), "A_Inputs")
b_path = join(main_path, "TP_0"+str(tp), "B_Outputs")
c_path = join(main_path, "TP_0"+str(tp), "C_Programs")
d_path = join(main_path, "TP_0"+str(tp), "D_LaTeX")

folders = [a_path, b_path, c_path, d_path]

for folder in folders:
    makedirs(folder, exist_ok=True)

# ============================================================
# 3. PRINCIPALES OUTPUTS
# ============================================================

b_path_data = join(b_path, "B_Data")
b_path_figr = join(b_path, "B_Figure")
b_path_tabl = join(b_path, "B_Table")
b_path_text = join(b_path, "B_Text")

folders = [b_path_data, b_path_figr, b_path_tabl, b_path_text]

for folder in folders:
    makedirs(folder, exist_ok=True)
```
